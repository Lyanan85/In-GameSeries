## 2. Data de Solicitação e Status de Fechamento

- **Data solicitada para fechar a agenda** (representada por `data_solicitacao`) refere-se ao momento em que a agenda foi **fechada** (solicitada) para um atendimento.
- **Status de fechamento** (representado pelas colunas `fechamento_s4`, `fechamento_siac` e `fechamento_hapvida`) indica a data em que o atendimento efetivamente deveria ocorrer, ou seja, o **dia do fechamento real**, quando o profissional ou substituto deveria estar presente.
