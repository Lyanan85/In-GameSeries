```python
import pandas as pd
```

```python
df=pd.read_csv("/content/imigrantes_canada.csv")
df
```

```python
df.info()
```

```python
df.set_index("País",inplace=True)
df
```

```python
anos=list(map(str,range(1980,2014)))
anos
```

```python
brasil=df.loc["Brasil",anos]
brasil
```

```python
brasil_dict={
    "ano":brasil.index.tolist(),
    "imigrantes":brasil.values.tolist()
}
dados_brasil=pd.DataFrame(brasil_dict)
display(dados_brasil)
print("-.-"*50)
print(brasil_dict)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))
plt.title("Imigração do Brasil para o Canadá",fontsize=12,loc="left")
plt.xlabel("Ano")
plt.ylabel("Número de imigrantes")
plt.plot(dados_brasil["ano"],dados_brasil["imigrantes"])
plt.xticks(["1980","1985","1990","1995","2000","2005","2010"])
plt.yticks([500,1000,1500,2000,2500,3000])
plt.show()
```

```python
argentina_brasil=df.loc[["Brasil","Argentina"],anos]
argentina_brasil
```

```python
df_comparacao=argentina_brasil.T
df_comparacao.head()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df_comparacao["Brasil"],label="Brasil")
plt.plot(df_comparacao["Argentina"],label="Argentina")
plt.title("Imigração do Brasil e Argentina para o Canadá",fontsize=12,loc="left")
plt.xlabel("Ano")
plt.ylabel("Número de imigrantes")
plt.xticks(["1980","1985","1990","1995","2000","2005","2010"])
plt.yticks([500,1000,1500,2000,2500,3000])
plt.legend()
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(dados_brasil["ano"],dados_brasil["imigrantes"])
ax.set_title("Imigraçaõ do Brasil para o Canadá\n1980 a 2013",fontsize=12,loc="center")
ax.set_xlabel("Ano")
ax.set_ylabel("Número de Imigrantes")
ax.set_frame_on(False)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.show()
```

```python
fig,axs=plt.subplots(1,2,figsize=(15,5))
axs[0].plot(dados_brasil["ano"],dados_brasil["imigrantes"])
axs[0].set_title("Imigração do Brasil para o Canadá\n1980 a 2013",fontsize=12)
axs[0].set_xlabel("Ano")
axs[0].set_ylabel("Número de Imigrantes")
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()


axs[1].boxplot(dados_brasil["imigrantes"])
axs[1].set_title("Boxplot da Imigração do Brasil para o Canadá\n1980 a 2023")
axs[1].set_xlabel("Brasil")
axs[1].set_ylabel("Imigrantes")
axs[1].grid()
plt.show()
```

```python
fig,axs=plt.subplots(2,2,figsize=(15,10))
fig.subplots_adjust(hspace=0.5,wspace=0.3)

axs[0,0].plot(df.loc["Brasil",anos])
axs[0,0].set_title("Brasil")

axs[0,1].plot(df.loc["Colômbia",anos])
axs[0,1].set_title("Colômbia")

axs[1,0].plot(df.loc["Argentina",anos])
axs[1,0].set_title("Argentina")

axs[1,1].plot(df.loc["Peru",anos])
axs[1,1].set_title("Peru")

for ax in axs.flat:
  ax.set_xlabel("Ano")
  ax.set_ylabel("Número de Imigrantes")
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.grid()
ymin=0
ymax=7000
for ax in axs.ravel():
  ax.set_ylim(ymin,ymax)


plt.show()
```
