```python
DF["station_antwerp"]=DF["station_antwerp"].interpolate(method="linear")
DF["station_paris"]=DF["station_paris"].interpolate(method="linear")
DF["station_london"]=DF["station_london"].interpolate(method="linear")
DF
```

```python
DF['hour'] = pd.to_datetime(DF['datetime']).dt.hour

# Calcular a média por hora do dia
hourly_mean = DF.groupby('hour')['station_antwerp'].transform('mean')

# Substituir os valores nulos pela média do horário
DF['station_antwerp'] = DF['station_antwerp'].fillna(hourly_mean)
DF["station_paris"]=DF["station_paris"].fillna(hourly_mean)
DF["station_london"]=DF["station_london"].fillna(hourly_mean)
display(DF)
```

```python
plt.figure(figsize=(10,6))
plt.plot(DF["datetime"], DF["station_antwerp"], label="Antuérpia", color=laranja)
plt.plot(DF["datetime"],DF["station_paris"],label="Paris", color=ciano)
plt.plot(DF["datetime"], DF["station_london"], label="Londres", color=lilas)
plt.legend()
plt.xlabel("Data")
plt.ylabel("Concentração de NO₂ (µg/m³)")
plt.title("Concentração de NO₂ nas Estações após interpolação")
plt.show()
```

# import matplotlib.dates as mdates: Importa o módulo mdates do Matplotlib, que contém ferramentas específicas para formatar e manipular datas em gráficos.


```python
import seaborn as sns
import matplotlib.dates as mdates
# Criando subplots para os histogramas
figura, axes = plt.subplots(1, 3, figsize=(24, 8))
# Histograma 1: Antuérpia
sns.histplot(data=DF, x="datetime", weights="station_antwerp", kde=True, ax=axes[0], color=laranja, bins=30, fill=True, linewidth=2, alpha=0.5)
axes[0].set_title("Distribuição de NO₂ em Antuérpia")
axes[0].set_xlabel("Data")
axes[0].set_ylabel("Concentração de NO₂ (µg/m³)")
axes[0].xaxis.set_major_locator(mdates.MonthLocator(interval=1))  # Exibe meses a cada 1
axes[0].xaxis.set_major_formatter(mdates.DateFormatter("%b-%Y"))  # Formato mês-ano
axes[0].tick_params(axis='x', rotation=45)  # Rotacionando as labels
axes[0].grid(True)
# Histograma 2: Paris
sns.histplot(data=DF, x="datetime", weights="station_paris", kde=True, ax=axes[1], color=ciano, bins=30, fill=True, linewidth=2, alpha=0.5)
axes[1].set_title("Distribuição de NO₂ em Paris")
axes[1].set_xlabel("Data")
axes[1].set_ylabel("Concentração de NO₂ (µg/m³)")
axes[1].xaxis.set_major_locator(mdates.MonthLocator(interval=1))
axes[1].xaxis.set_major_formatter(mdates.DateFormatter("%b-%Y"))
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True)
# Histograma 3: Londres
sns.histplot(data=DF, x="datetime", weights="station_london", kde=True, ax=axes[2], color=lilas, bins=30, fill=True, linewidth=2, alpha=0.5)
axes[2].set_title("Distribuição de NO₂ em Londres")
axes[2].set_xlabel("Data")
axes[2].set_ylabel("Concentração de NO₂ (µg/m³)")
axes[2].xaxis.set_major_locator(mdates.MonthLocator(interval=1))
axes[2].xaxis.set_major_formatter(mdates.DateFormatter("%b-%Y"))
axes[2].tick_params(axis='x', rotation=45)
axes[2].grid(True)
# Ajuste do layout e espaçamento
plt.tight_layout(pad=2.0)
plt.show()
```
