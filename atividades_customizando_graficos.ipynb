#**Criando gráficos customizados**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

###**Atividade 1**

```python
df_aluguel= pd.read_csv('https://raw.githubusercontent.com/alura-cursos/python_dados/refs/heads/main/Dados/apartamentos_aluguel.csv')
```

```python
df_aluguel.head()
```

```python
# Agrupando por região e obtendo a média do valor
valor_regiao = df_aluguel.groupby('Regiao')['Valor'].mean().reset_index()

# Ordenando do maior para o menor valor
valor_regiao_ordernado = valor_regiao.sort_values(by=['Valor'], ascending=False)
```

```python
# Criando a figura e os eixos do gráfico
fig, ax = plt.subplots(figsize=(10,5))

# Ajustando o layout da figura para deixar mais espaço à direita
fig.subplots_adjust(right=0.7)

# Criando o gráfico de colunas
ax = sns.barplot(x='Regiao', y='Valor', data=valor_regiao_ordernado)

# Adicionando título ao gráfico
ax.set_title('Valor médio dos aluguéis de apartamentos por região em São Paulo', pad=30, fontsize=18, loc='left')

# Exibindo o gráfico
plt.show()
```

###**Atividade 6**

```python
df_bicicletas = pd.read_csv('https://raw.githubusercontent.com/alura-cursos/python_dados/refs/heads/main/Dados/contagem_bicicletas.csv')
```

```python
df_bicicletas.head()
```

```python
df_bicicletas['data_hora'] = pd.to_datetime(df_bicicletas['data_hora'])
df_bicicletas['data'] = df_bicicletas['data_hora'].dt.date
df_bicicletas.head()
```

```python
contagem_bicicletas_data = df_bicicletas.groupby('data')['contagem'].sum().reset_index()
contagem_bicicletas_data
```

###**Atividade 7**

```python
# Criando a figura e os eixos do gráfico
fig, ax = plt.subplots(figsize=(14,5))

# Ajustando o layout da figura para deixar mais espaço à direita
fig.subplots_adjust(right=0.7)

# Criando o gráfico de linhas
ax = sns.lineplot(data=contagem_bicicletas_data, x='data', y='contagem')

# Adicionando título ao gráfico
ax.set_title('Demanda diária por bicicletas compartilhadas em Londres', pad=70, fontsize=18, loc='left')

# Ocultando o rótulo do eixo X
ax.set_xlabel('')

# Alterando o rótulo do eixo Y
ax.set_ylabel('Número de bicicletas')

# Exibindo o gráfico
plt.show()
```
