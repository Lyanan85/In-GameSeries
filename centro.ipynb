```python
mapeamento_dentistas = {
    "LUCIANE ALVES NUNES":"LUCIANE ALVES NUNES"}
# Aplicar o mapeamento
df['Dentista'] = df['Dentista'].str.strip().str.lower().replace(mapeamento_dentistas).str.title()

df.to_csv("df_inicial.csv",index=False)
```

```python
df["Orçamento"] = pd.to_datetime(df["Orçamento"], format="%d/%m/%Y %H:%M", errors="coerce")
def dia_da_semana(data):
    dias = {
        0: "Segunda-feira",
        1: "Terça-feira",
        2: "Quarta-feira",
        3: "Quinta-feira",
        4: "Sexta-feira",
        5: "Sábado",
        6: "Domingo"
    }
    if pd.notnull(data):
        return dias[data.weekday()]
    return None
df["Dia_semana"]=df["Orçamento"].apply(dia_da_semana)
df
```

```python
import re
def extrair_turno(data_srt):
    if pd.isnull(data_srt):
        return "Sem Horário"

    hora = data_srt.hour  # pega a hora direto do datetime

    if 5 <= hora < 12:
        return "Manhã"
    elif 12 <= hora < 18:
        return "Tarde"
    elif 0 <= hora <= 23:
        return "Noite"
    else:
        return "Sem Horário"


    # Define o turno
    if 5 <= hora < 12:
        return "Manhã"
    elif 12 <= hora < 18:
        return "Tarde"
    elif 0 <= hora <= 23:
        return "Noite"
    else:
        return "Sem Horário"

df["Turno"]=df["Orçamento"].apply(extrair_turno)
df
```

```python
def calcular_valor_vendido(row):
    if row['Total'] == row['Pago'] and row['Devedor'] == 0:
        return row['Total']
    elif row['Devedor'] != 0:
        return row['Pago']
    else:
        return 0
```
